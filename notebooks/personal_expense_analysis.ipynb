# Personal Expense Tracker Analysis

A Pandas, NumPy and Matplotlib walkthrough of the 2025 expense dataset.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid')
df = pd.read_csv('../data/personal_expenses_2025.csv', parse_dates=['date'])
df.head()
```

## 1. Spending by category


```python
by_category = df.groupby('category').agg(
    total_amount=('amount_bdt', 'sum'),
    average_amount=('amount_bdt', 'mean'),
    transaction_count=('amount_bdt', 'count')
).round(2).sort_values('total_amount', ascending=False)
by_category['share_pct'] = (by_category['total_amount'] / df['amount_bdt'].sum() * 100).round(2)
by_category
```

## 2. Monthly spending line plot


```python
monthly = df.groupby('month')['amount_bdt'].sum().reindex(range(1, 13), fill_value=0)

plt.figure(figsize=(10, 5))
plt.plot(monthly.index, monthly.values, marker='o', color='#2f81f7')
plt.title('Monthly Spending 2025')
plt.xlabel('Month')
plt.ylabel('Total expenses (BDT)')
plt.xticks(range(1, 13))
plt.show()
```
